*2.3 Valores nulos*

Primero uso el método pedido en la consigna (`.isnull().sum()`) y después lo paso a porcentaje para dimensionar el impacto de cada columna.

```python
# Cuento los nulos por columna y me quedo solo con las que tienen alguno
nulos = df.isnull().sum()
nulos[nulos > 0]
```

```python
def resumen_nulos(dataframe):
    # Creo un diccionario vacío donde guardo el porcentaje de nulos de cada columna
    porcentajes = {}
    # Guardo la cantidad total de filas
    total_filas = len(dataframe)

    # Recorro cada columna del DataFrame
    for columna in dataframe.columns:
        # Cuento los nulos de la columna
        cantidad = dataframe[columna].isnull().sum()
        # Solo registro las columnas que tienen al menos un nulo
        if cantidad > 0:
            # Calculo el porcentaje y lo guardo en el diccionario
            porcentajes[columna] = round(cantidad / total_filas * 100, 2)

    # Convierto el diccionario en una Serie y la ordeno de mayor a menor
    return pd.Series(porcentajes, name='% nulos').sort_values(ascending=False)

resumen_nulos(df)
```

**Observaciones:**
- `vida_util` (97,87%) y `observaciones` (94,30%) están casi vacías: no tiene sentido imputarlas, conviene eliminarlas.
- El resto de las columnas con nulos tiene menos del 0,25% de faltantes, así que puedo tratarlas con imputación o eliminación puntual sin afectar el análisis.

####**3. Limpieza rigurosa**

Aplico las transformaciones en un orden lógico: primero elimino lo que no aporta información, después corrijo los tipos (para que los cálculos posteriores sean confiables), luego trato nulos y valores inválidos y por último, verifico duplicados y reseteo el índice.

*3.1 Eliminación de columnas sin información útil*

*3.2 Conversión de tipos de datos*

```python
# Convierto las fechas de texto a datetime
df['fecha_data'] = pd.to_datetime(df['fecha_data'], format='%Y-%m-%d')
# fechaingreso tiene fracciones de segundo de largo variable, por eso uso el formato ISO8601
df['fechaingreso'] = pd.to_datetime(df['fechaingreso'], format='ISO8601')

# Convierto rectificado de 't'/'f' a booleano con un diccionario de equivalencias
df['rectificado'] = df['rectificado'].map({'t': True, 'f': False})

# Convierto a categoría las columnas de texto con pocos valores distintos que se repiten mucho
columnas_categoricas = ['idempresa', 'empresa', 'tipoextraccion', 'tipoestado', 'tipopozo',
                        'formprod', 'formacion', 'cuenca', 'provincia', 'tipo_de_recurso',
                        'proyecto', 'clasificacion', 'subclasificacion', 'sub_tipo_recurso']
for columna in columnas_categoricas:
    df[columna] = df[columna].astype('category')

# Verifico los tipos de las columnas que modifiqué
df[['fecha_data', 'fechaingreso', 'rectificado', 'tipopozo', 'formacion']].dtypes
```

La conversión de tipos no es solo una cuestión de prolijidad: además de permitir operar con fechas (por ejemplo, obtener los días de cada mes en el paso 3.7), reduce la memoria del DataFrame de 635,6 MB a 182,0 MB (una reducción de más del 70%).

*3.3 Imputación por moda del pozo: `tipopozo`, `tipoextraccion`, `tipoestado`*

Estas tres columnas tienen nulos en las mismas filas. En lugar de imputar con la moda general del dataset (que asignaría "Gasífero" a cualquier pozo), uso la moda de cada pozo: si un pozo tiene otros meses informados, completo con el valor que más se repite en su propio historial.

```python
# Verifico que los nulos de las tres columnas caen en las mismas filas
nulos_tipopozo = df['tipopozo'].isnull()
print('Coinciden con tipoextraccion:', (nulos_tipopozo == df['tipoextraccion'].isnull()).all())
print('Coinciden con tipoestado:', (nulos_tipopozo == df['tipoestado'].isnull()).all())
```

```python
def moda_por_pozo(dataframe, columna):
    # Creo un diccionario donde cada pozo tiene su propio contador de valores
    contadores = {}

    # Recorro al mismo tiempo el id del pozo y el valor de la columna, fila por fila
    for pozo, valor in zip(dataframe['idpozo'], dataframe[columna]):
        # Salteo los valores nulos porque no aportan a la moda
        if pd.isnull(valor):
            continue
        # Si es la primera vez que veo este pozo, le creo un contador vacío
        if pozo not in contadores:
            contadores[pozo] = {}
        # Si es la primera vez que veo este valor en el pozo, lo inicializo en cero
        if valor not in contadores[pozo]:
            contadores[pozo][valor] = 0
        # Actualizo el contador
        contadores[pozo][valor] += 1

    # Creo el diccionario final con la moda de cada pozo
    modas = {}
    # Recorro cada pozo y su contador
    for pozo, conteo in contadores.items():
        # Me quedo con el valor que tiene el conteo más alto
        modas[pozo] = max(conteo, key=conteo.get)

    return modas
```

```python
# Recorro las tres columnas y completo sus nulos con la moda del pozo
for columna in ['tipopozo', 'tipoextraccion', 'tipoestado']:
    nulos_antes = df[columna].isnull().sum()
    # Calculo la moda de cada pozo para esta columna
    modas = moda_por_pozo(df, columna)
    # Busco la moda correspondiente a cada fila según su pozo
    valores_imputados = df['idpozo'].map(modas)
    # Completo solo los nulos; paso a texto para poder combinar con la columna categórica
    df[columna] = df[columna].astype('object').fillna(valores_imputados).astype('category')
    nulos_despues = df[columna].isnull().sum()
    print(f'{columna}: {nulos_antes} nulos antes -> {nulos_despues} nulos después')
```

Con la moda del pozo recuperé 610 de las 614 filas con nulos. Las 4 restantes pertenecen a pozos que no tienen ningún mes con estos datos informados, así que no hay de dónde imputar: las elimino en el paso siguiente.

*3.4 Eliminación estratégica de filas con nulos no recuperables*

Las filas que quedaron sin `tipopozo` y las que no tienen `clasificacion`, `subclasificacion` o `sub_tipo_recurso` pertenecen a pozos que no tienen ningún mes informado en esas columnas, por lo que no hay de dónde imputar. Como representan una proporción mínima del total, las elimino.

```python
# Elimino las filas que siguen teniendo nulos en columnas categóricas clave
columnas_clave = ['tipopozo', 'tipoextraccion', 'tipoestado', 'clasificacion', 'subclasificacion', 'sub_tipo_recurso']
filas_antes = len(df)
df = df.dropna(subset=columnas_clave)
filas_eliminadas = filas_antes - len(df)

print('Filas eliminadas:', filas_eliminadas)
print(f'Porcentaje del total: {filas_eliminadas / filas_antes * 100:.2f}%')
```

*3.5 Eliminación de producciones negativas*

```python
# Me quedo solo con las filas donde las tres producciones son mayores o iguales a cero
filas_antes = len(df)
df = df[(df['prod_pet'] >= 0) & (df['prod_gas'] >= 0) & (df['prod_agua'] >= 0)]
print('Filas eliminadas por producción negativa:', filas_antes - len(df))
```

*3.6 Imputación de `profundidad` inválida con la mediana por formación*

Trato como dato faltante tanto las profundidades en 0 (el pozo no puede tener profundidad nula) como las mayores a 8.000 m: el valor que más se repite ahí es 378.939 m, físicamente imposible (los pozos de Vaca Muerta rondan los 2.500 a 4.000 m de profundidad vertical), lo que indica un error de carga.

Como la profundidad de un pozo depende sobre todo de la formación geológica a la que apunta, imputo con la mediana de su formación. Elijo la mediana en lugar de la media porque es resistente a valores extremos.

```python
# Reemplazo los valores inválidos por NaN para tratarlos como faltantes
df['profundidad'] = df['profundidad'].astype('float64')
invalidas = (df['profundidad'] == 0) | (df['profundidad'] > 8000)
df.loc[invalidas, 'profundidad'] = np.nan
print('Profundidades marcadas como faltantes:', df['profundidad'].isnull().sum())
```

```python
def mediana_por_formacion(dataframe):
    # Creo un diccionario donde guardo la lista de profundidades válidas de cada formación
    profundidades = {}

    # Recorro al mismo tiempo la formación y la profundidad de cada fila
    for formacion, profundidad in zip(dataframe['formacion'], dataframe['profundidad']):
        # Salteo las profundidades faltantes
        if pd.isnull(profundidad):
            continue
        # Si es la primera vez que veo la formación, le creo una lista vacía
        if formacion not in profundidades:
            profundidades[formacion] = []
        # Agrego la profundidad a la lista de su formación
        profundidades[formacion].append(profundidad)

    # Calculo la mediana de cada formación
    medianas = {}
    for formacion, lista in profundidades.items():
        medianas[formacion] = np.median(lista)

    return medianas

# Calculo las medianas y completo los faltantes según la formación de cada fila
medianas = mediana_por_formacion(df)
df['profundidad'] = df['profundidad'].fillna(df['formacion'].astype('object').map(medianas))
print('Profundidades faltantes después de imputar:', df['profundidad'].isnull().sum())
```

*3.7 Corrección del tiempo efectivo (`tef`)*

`tef` son los días del mes en que el pozo estuvo produciendo, así que no puede superar la cantidad de días de ese mes. Recorto los valores excedidos al máximo posible usando la fecha del registro.

```python
# Obtengo la cantidad de días de cada mes a partir de la fecha (ya convertida a datetime)
dias_del_mes = df['fecha_data'].dt.days_in_month
print('Registros con tef mayor a los días del mes:', (df['tef'] > dias_del_mes).sum())

# Recorto tef al máximo de días del mes
df['tef'] = df['tef'].clip(upper=dias_del_mes)
print('Después de corregir:', (df['tef'] > dias_del_mes).sum())
```

*3.8 Duplicados*

```python
# Elimino duplicados exactos y por clave (en este dataset no hay, pero dejo el paso para que el flujo sea reproducible si el archivo se actualiza)
df = df.drop_duplicates()
df = df.drop_duplicates(subset=['idpozo', 'anio', 'mes'])
print('Duplicados por clave pozo-año-mes:', df.duplicated(subset=['idpozo', 'anio', 'mes']).sum())
```

*Pregunta 1: ¿Cómo evolucionó la producción anual total de petróleo y gas no convencional?*

```python
# Agrupo por año y sumo la producción de petróleo y gas
produccion_anual = df.groupby('anio')[['prod_pet', 'prod_gas']].sum()

# Agrego la cantidad de pozos distintos que reportaron en cada año
produccion_anual['pozos_activos'] = df.groupby('anio')['idpozo'].nunique()
produccion_anual
```

**Interpretación:**

- Durante la primera década el gas es el producto principal en volumen; desde 2021 el petróleo acelera fuertemente y en 2025 prácticamente alcanza al gas en volumen (recordar que el gas está en miles de m³).
- 2020 es el único año en que el gas cae (efecto pandemia), mientras que el petróleo siguió creciendo.
- 2026 es un año incompleto (los datos llegan hasta julio), por eso su total es menor: no debe leerse como una caída.

*Pregunta 2: ¿Qué formaciones rinden más por pozo?*

Calculo la producción mensual promedio por pozo en cada formación, separando pozos petrolíferos (petróleo) y gasíferos (gas) y considerando solo los meses en extracción efectiva para no promediar meses en que el pozo estuvo parado. Además, dejo afuera las formaciones con menos de 10 pozos, porque un promedio calculado sobre 1 o 2 pozos no es representativo.

```python
# Me quedo con los meses en que el pozo estuvo produciendo
en_produccion = df[df['tipoestado'] == 'Extracción Efectiva']

# Pozos petrolíferos: promedio y cantidad de pozos por formación
petroliferos = en_produccion[en_produccion['tipopozo'] == 'Petrolífero']
rendimiento_pet = petroliferos.groupby('formacion', observed=True).agg(
    prod_pet_promedio=('prod_pet', 'mean'),
    cantidad_pozos=('idpozo', 'nunique'))
# Descarto las formaciones con menos de 10 pozos porque su promedio no es representativo
rendimiento_pet = rendimiento_pet[rendimiento_pet['cantidad_pozos'] >= 10]
rendimiento_pet.sort_values('prod_pet_promedio', ascending=False)
```

```python
# Pozos gasíferos: promedio y cantidad de pozos por formación
gasiferos = en_produccion[en_produccion['tipopozo'] == 'Gasífero']
rendimiento_gas = gasiferos.groupby('formacion', observed=True).agg(
    prod_gas_promedio=('prod_gas', 'mean'),
    cantidad_pozos=('idpozo', 'nunique'))
# Aplico el mismo criterio de al menos 10 pozos por formación
rendimiento_gas = rendimiento_gas[rendimiento_gas['cantidad_pozos'] >= 10]
rendimiento_gas.sort_values('prod_gas_promedio', ascending=False).head(6)
```

**Interpretación:**
- Vaca Muerta domina en ambos tipos de pozo: un pozo petrolífero produce en promedio unos 914 m³ de petróleo por mes, más de 7 veces lo que rinde la siguiente formación y concentra la gran mayoría de los pozos petrolíferos (2.513 pozos).
- En gas la diferencia es menor: Vaca Muerta lidera (unos 3.624 miles de m³/mes), seguida por Agrio y Punta Rosada. Lajas tiene muchos pozos gasíferos (581) pero con un rendimiento promedio mucho más bajo, consistente con ser un reservorio tight.
- Esto refuerza la decisión de analizar pozos petrolíferos y gasíferos por separado.

*Pregunta 3: ¿Qué empresas concentran la producción de petróleo?*

```python
# Sumo el petróleo producido por cada empresa en todo el período
petroleo_por_empresa = df.groupby('empresa', observed=True)['prod_pet'].sum().sort_values(ascending=False)

# Calculo qué porcentaje del total representa cada empresa
participacion = (petroleo_por_empresa / petroleo_por_empresa.sum() * 100).round(2)

top_empresas = pd.DataFrame({'prod_pet_total': petroleo_por_empresa, 'participacion_%': participacion})
top_empresas.head(10)
```

**Interpretación:**
- La producción de petróleo no convencional está muy concentrada: YPF sola aporta el 58,2% del total histórico y las 5 primeras empresas superan el 84%.
- Vista aparece con dos razones sociales distintas ("VISTA ENERGY ARGENTINA SAU" y "VISTA OIL & GAS ARGENTINA SAU"), probablemente por un cambio de denominación. Para un análisis por grupo empresario convendría unificarlas; acá las dejo como figuran en la fuente.

*Pregunta 4: ¿Cuántos pozos hay por provincia y de qué tipo de recurso (shale o tight)?*

```python
# Cuento los pozos distintos por provincia y sub tipo de recurso
pozos_por_provincia = df.groupby(['provincia', 'sub_tipo_recurso'], observed=True)['idpozo'].nunique()

# Paso el sub tipo de recurso a columnas para leerlo como tabla cruzada
pozos_por_provincia = pozos_por_provincia.unstack(fill_value=0)
pozos_por_provincia['total'] = pozos_por_provincia.sum(axis=1)
pozos_por_provincia.sort_values('total', ascending=False)
```

**Interpretación:**
- Neuquén concentra el 90% de los pozos no convencionales del país y es la única provincia donde predomina el shale (3.420 pozos, casi todos de Vaca Muerta).
- En Río Negro y Santa Cruz el desarrollo es principalmente tight.
- Mendoza, Chubut y Salta tienen una presencia todavía marginal.
